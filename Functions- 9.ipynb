## Functions in Python

A function is a reusable block of code that performs a specific task. Instead of writing the same code again and again, you define it once and call it whenever needed.

1. **def** - keyword used to define a function
2. **parameters** - values passed into a function
3. **return** - sends a value back from the function
4. **arguments** - actual values given when calling the function

## Key things to remember

- A function only runs when it is **called**, not when it is defined
- `return` sends a value back and immediately ends the function; `print()` just displays something
- Default arguments must come **after** non-default arguments in the function definition
- `*args` collects extra positional arguments into a tuple; `**kwargs` collects extra keyword arguments into a dictionary
- Variables created inside a function are **local** and don't exist outside it, unless declared `global`

## Basic Functions

```python
# Basic function with no parameters
def greet():
    print("Hello, welcome!")

greet()
```

```python
# Function with a parameter
def greet(name):
    print(f"Hello, {name}!")

greet("Shanthi")
```

```python
# Function with multiple parameters
def add(a, b):
    print(a + b)

add(5, 10)
```

```python
# Function with a return value
def add(a, b):
    return a + b

result = add(5, 10)
print(result)
```

```python
# Function returning multiple values (as a tuple)
def get_min_max(numbers):
    return min(numbers), max(numbers)

low, high = get_min_max([4, 1, 9, 2, 7])
print(f"Min: {low}, Max: {high}")
```

## Default, Keyword, and Variable-length Arguments

```python
# Default parameter value
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()          # uses default
greet("Aman")    # overrides default
```

```python
# Keyword arguments - order doesn't matter when named
def student_info(name, age):
    print(f"Name: {name}, Age: {age}")

student_info(age=22, name="Riya")
```

```python
# *args - accepts any number of positional arguments
def total_sum(*numbers):
    return sum(numbers)

print(total_sum(1, 2, 3))
print(total_sum(10, 20, 30, 40))
```

```python
# **kwargs - accepts any number of keyword arguments
def student_profile(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

student_profile(name="Riya", age=22, city="Hyderabad")
```

```python
# Combining normal, *args, and **kwargs in one function
def order_summary(customer, *items, **info):
    print(f"Customer: {customer}")
    print("Items:", items)
    print("Extra info:", info)

order_summary("Riya", "Notebook", "Pen", payment="UPI", express=True)
```

## Scope, Lambda, and Special Function Concepts

```python
# Local vs global variable scope
x = 10   # global variable

def show_number():
    x = 5   # local variable, different from global x
    print("Inside function:", x)

show_number()
print("Outside function:", x)
```

```python
# Using the global keyword to modify a global variable
count = 0

def increment():
    global count
    count += 1

increment()
increment()
print(count)
```

```python
# Lambda function - a small anonymous one-line function
square = lambda x: x ** 2
print(square(5))
```

```python
# Lambda used inside sorted() with a key
students = [("Riya", 88), ("Aman", 95), ("Kabir", 76)]
sorted_by_marks = sorted(students, key=lambda x: x[1], reverse=True)
print(sorted_by_marks)
```

```python
# Function as an argument to another function
def apply_operation(a, b, operation):
    return operation(a, b)

def multiply(x, y):
    return x * y

print(apply_operation(5, 3, multiply))
```

```python
# Recursive function - calls itself (factorial example)
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

```python
# Docstring - documenting what a function does
def add(a, b):
    """Returns the sum of two numbers."""
    return a + b

print(add(4, 6))
print(add.__doc__)
```

```python
# Function with a mutable default argument (a common gotcha)
def add_item(item, cart=None):
    if cart is None:
        cart = []
    cart.append(item)
    return cart

print(add_item("apple"))
print(add_item("banana"))   # new list each time, not shared
```

## Bigger Examples

```python
# Simple calculator using functions
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: division by zero"
    return a / b

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

if operator == "+":
    print("Result:", add(num1, num2))
elif operator == "-":
    print("Result:", subtract(num1, num2))
elif operator == "*":
    print("Result:", multiply(num1, num2))
elif operator == "/":
    print("Result:", divide(num1, num2))
else:
    print("Invalid operator")
```

```python
# 20. Prime number checker as a reusable function
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

for n in range(1, 20):
    if is_prime(n):
        print(n, "is prime")
```

```python
# 21. Password validator function
def is_strong_password(password):
    has_digit = any(char.isdigit() for char in password)
    has_upper = any(char.isupper() for char in password)
    long_enough = len(password) >= 8
    return has_digit and has_upper and long_enough

pwd = input("Enter a password: ")
if is_strong_password(pwd):
    print("Strong password")
else:
    print("Weak password")
```

```python
# Temperature converter using functions
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5/9

print(celsius_to_fahrenheit(37))
print(fahrenheit_to_celsius(98.6))
```

```python
# Student grade calculator using a function
def calculate_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return "F"

students = {"Riya": 92, "Aman": 68, "Kabir": 45}
for name, marks in students.items():
    print(f"{name}: {calculate_grade(marks)}")
```

```python
# ATM withdrawal function with balance check
def withdraw(balance, amount):
    if amount > balance:
        return balance, "Insufficient funds"
    else:
        balance -= amount
        return balance, "Withdrawal successful"

balance = 5000
balance, message = withdraw(balance, 2000)
print(message, "| New balance:", balance)
```

```python
# Function that returns a function (closures)
def multiplier(factor):
    def multiply_by(number):
        return number * factor
    return multiply_by

double = multiplier(2)
triple = multiplier(3)

print(double(5))
print(triple(5))
```
