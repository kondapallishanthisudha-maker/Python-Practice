## Modules in Python

A module is simply a Python file (`.py`) containing functions, variables, and classes that can be reused in other programs. Instead of rewriting code, you **import** a module and use what's inside it.

1. **Built-in modules** - come pre-installed with Python (e.g. `math`, `random`, `datetime`)
2. **User-defined modules** - Python files you create yourself and import
3. **Third-party modules** - installed separately using `pip` (e.g. `numpy`, `pandas`)
4. **Package** - a folder containing multiple related modules, with an `__init__.py` file

## Importing Built-in Modules

````python
# Importing an entire module
import math

print(math.sqrt(25))
print(math.pi)
````

````python
# Importing a module with an alias (shortened name)
import math as m

print(m.factorial(5))
print(m.pow(2, 3))
````

````python
# Importing specific functions from a module
from math import sqrt, ceil, floor

print(sqrt(16))
print(ceil(4.3))
print(floor(4.7))
````

````python
# Importing everything from a module using * (not usually recommended)
from math import *

print(sin(0))
print(cos(0))
````

````python
# Viewing everything available inside a module using dir()
import math
print(dir(math)[:10])   # first 10 items just to preview
````

## The random Module

````python
# random.randint() - random whole number in a range (inclusive)
import random

print(random.randint(1, 10))
````

````python
# random.choice() - pick a random item from a list
import random

fruits = ["apple", "banana", "cherry"]
print(random.choice(fruits))
````

````python
#random.shuffle() - shuffle a list in place
import random

cards = [1, 2, 3, 4, 5]
random.shuffle(cards)
print(cards)
````

## The datetime Module

````python
# Getting the current date and time
from datetime import datetime

now = datetime.now()
print(now)
````

````python
# Formatting a date using strftime()
from datetime import datetime

now = datetime.now()
print(now.strftime("%d-%m-%Y"))
print(now.strftime("%H:%M:%S"))
````

## Other Useful Built-in Modules

````python
# os module - checking the current working directory
import os

print(os.getcwd())
````

````python
# sys module - checking the Python version being used
import sys

print(sys.version)
````

## Creating Your Own Module (User-defined Module)

In a real project, you would save the function below into a separate file called `mymodule.py`, then import it in another file using `import mymodule`. Since a notebook works a little differently, we define the function in one cell and use it in the next cell to simulate the same idea.

````python
#  Simulating a user-defined module
# --- Imagine this part is saved in a file called mymodule.py ---
def greet(name):
    return f"Hello, {name}! Welcome to Python."

def add(a, b):
    return a + b

PI = 3.14159
````

````python
# --- Imagine this part is in another file that does: import mymodule ---
print(greet("Riya"))
print(add(5, 10))
print(PI)
````

**Real file example (for your own practice on disk):**
```python
# mymodule.py

    return f"Hello, {name}!"

# main.py
import mymodule
print(mymodule.greet("Riya"))
```

## The `__name__ == "__main__"` Pattern

````python
# Checking __name__ - useful so a module's test code only runs when executed directly, not when imported
def main():
    print("This runs only when the file is executed directly, not when imported")

if __name__ == "__main__":
    main()
````

## Functions Inside Modules -

````python
# Using multiple functions together from the math module
import math

def circle_area(radius):
    return math.pi * radius ** 2

def circle_circumference(radius):
    return 2 * math.pi * radius

r = 5
print(f"Area: {circle_area(r):.2f}")
print(f"Circumference: {circle_circumference(r):.2f}")
````

````python
# Combining random and a custom function - simple dice roll simulator
import random

def roll_dice():
    return random.randint(1, 6)

for _ in range(5):
    print(roll_dice())
````

## Packages (Modules Grouped Together)

A package is just a folder of related modules. For example:
```
mypackage/
    __init__.py
    shapes.py
    conversions.py
```
You would then import from it like:
```python
from mypackage import shapes
shapes.circle_area(5)
```

````python
#Example of importing from a real installed package's submodule
from collections import Counter

votes = ["A", "B", "A", "C", "B", "A"]
result = Counter(votes)
print(result)
````

## Bigger Examples

````python
# Bigger: temperature converter tool using the math module for rounding
import math

def celsius_to_fahrenheit(c):
    return math.floor((c * 9/5) + 32)

temps = [0, 20, 37, 100]
for t in temps:
    print(f"{t}°C = {celsius_to_fahrenheit(t)}°F")
````

````python
# Bigger: password generator using random and string modules
import random
import string

def generate_password(length=8):
    characters = string.ascii_letters + string.digits
    return "".join(random.choice(characters) for _ in range(length))

print(generate_password())
print(generate_password(12))
````

````python
# Bigger: log entry generator using datetime
from datetime import datetime

def log_message(message):
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    return f"[{timestamp}] {message}"

print(log_message("User logged in"))
print(log_message("File saved successfully"))
````

````python
# Bigger: student statistics report using the statistics module
import statistics

def generate_report(scores):
    print(f"Scores: {scores}")
    print(f"Mean: {statistics.mean(scores):.2f}")
    print(f"Median: {statistics.median(scores)}")
    print(f"Standard Deviation: {statistics.stdev(scores):.2f}")

generate_report([88, 92, 76, 65, 95, 70])
````
