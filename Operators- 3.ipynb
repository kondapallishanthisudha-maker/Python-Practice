```python
a = 12
b = 10
print(a | b)   # OR
```

```python
print(bin(12))
print(bin(10))
```

```python
a = 12
b = 10
print(a ^ b) #XOR
```

```python
a = 12
b = 10
bin_a = bin(a)[2:]
bin_b = bin(b)[2:]
and_result = a & b
bin_and = bin(and_result)[2:]

print(f"12 in binary: {bin_a}")
print(f"10 in binary: {bin_b}")
print(f"AND result: {and_result} in decimal, {bin_and} in binary")
```

```python
a = 12
b = 10
bin_a = bin(a)
bin_b = bin(b)
and_result = a | b
bin_and = bin(and_result)

print(f"12 in binary: {bin_a}")
print(f"10 in binary: {bin_b}")
print(f"AND result: {and_result} in decimal, {bin_and} in binary")
```

7. Identity Operators

```python
a = [1, 2, 3]
b = a
print(a is b) #is checks whether two variables point to the same object in memory
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a is b) # False - different objects
print(a == b) # True - same values 
```

```python
a = 10
b = 10
print(a is b) # True - small ints are cached by python
```

```python
result = None 
print(result is None) # correct, Pythonic way 
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
a = b
print(a is b)
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a is not b)
```

```python
def find_user(user_id):
    users = {1: "Riya", 2:"Aman"}
    return users.get(user_id) # returns None if not found

user = find_user(5)
if user is None:
    print("User Not Return")
else:
    print(f"User: {user}")      
```

```python
original = [1,2,3] 
Reference = original 
copy_of_it = original.copy()

print(original is Reference) # is only checks if two variables refer to the exact same obj
print(original is copy_of_it)
print(original == copy_of_it) # ' == ' compare the values
```
