## Packages in Python

A package is a **folder that groups related modules together**. It lets you organize a large project into logical sections instead of keeping every function in one giant file.

**Structure of a simple package:**
```
mypackage/
    __init__.py          <- marks the folder as a package
    shapes.py             <- a module inside the package
    conversions.py        <- another module inside the package
```

- **Module** = a single `.py` file
- **Package** = a folder containing multiple modules (and an `__init__.py` file)
- **Library** = a collection of packages (e.g. NumPy, Pandas)

Since a notebook is a single file, most examples below **simulate** how a package would work, and also show real installed packages you can actually import and run.

## Key things to remember

- A **package** is a folder of related modules; a **module** is a single `.py` file
- `__init__.py` marks a folder as a package and can control what gets exposed on import
- Built-in packages (like `os`, `collections`, `json`) are ready to use with no installation
- Third-party packages (like `numpy`, `pandas`, `requests`) must be installed first using `pip install package_name`
- Packages can contain **sub-packages**, creating a nested folder structure for large projects
- Use **absolute imports** for clarity in most projects; **relative imports** (`.` and `..`) only work from inside an actual package

## The role of `__init__.py`

`__init__.py` is a (often empty) file that tells Python "this folder is a package." It can also be used to control what gets imported automatically.

```python
# mypackage/__init__.py
from .shapes import circle_area
from .conversions import km_to_miles
```

With this, you could then simply do:
```python
import mypackage
mypackage.circle_area(5)      # works directly, without .shapes
```

## Sub-packages (Packages Inside Packages)

A package can also contain other packages, creating a nested structure:
```
mypackage/
    __init__.py
    shapes.py
    utils/
        __init__.py
        validators.py
        formatters.py
```
You would import from a sub-package like this:
```python
from mypackage.utils import validators
validators.is_valid_email("test@test.com")
```

````python
# Simulating a sub-package's validators module
# --- Imagine this is mypackage/utils/validators.py ---
def is_valid_email(email):
    return "@" in email and "." in email

def is_valid_phone(phone):
    return phone.isdigit() and len(phone) == 10
````

````python
# Simulating usage: from mypackage.utils import validators
print(is_valid_email("test@example.com"))
print(is_valid_phone("9876543210"))
````

## Relative vs Absolute Imports (Reference)

Inside real package files, you'll see two import styles:

```python
# Absolute import - full path from the project root
from mypackage.shapes import circle_area

# Relative import - relative to the current file's location (used INSIDE a package)
from .shapes import circle_area          # same folder
from ..otherpackage import something     # one folder up
```
Relative imports only work inside an actual package structure - they will not run inside a single notebook file.

## Bigger Examples

````python
# Bigger: combining functions from two "package modules" for a real task
# --- shapes.py + conversions.py combined together ---
def circle_area(radius):
    return 3.14159 * radius ** 2

def sqm_to_sqft(area_sqm):
    return area_sqm * 10.7639

radius = 4
area = circle_area(radius)
area_in_sqft = sqm_to_sqft(area)
print(f"Area: {area:.2f} sqm = {area_in_sqft:.2f} sqft")
````

````python
# 1Bigger: using multiple real built-in packages together in one small tool
import random
import string
from datetime import datetime

def generate_ticket_id():
    letters = "".join(random.choices(string.ascii_uppercase, k=3))
    numbers = "".join(random.choices(string.digits, k=4))
    timestamp = datetime.now().strftime("%Y%m%d")
    return f"{letters}-{numbers}-{timestamp}"

print(generate_ticket_id())
````

````python
# validating a small user registration form using "package-style" helper functions
def is_valid_email(email):
    return "@" in email and "." in email

def is_valid_phone(phone):
    return phone.isdigit() and len(phone) == 10

def is_strong_password(password):
    return len(password) >= 8 and any(c.isdigit() for c in password)

def register_user(email, phone, password):
    errors = []
    if not is_valid_email(email):
        errors.append("Invalid email")
    if not is_valid_phone(phone):
        errors.append("Invalid phone number")
    if not is_strong_password(password):
        errors.append("Weak password")

    if errors:
        return f"Registration failed: {errors}"
    return "Registration successful"

print(register_user("test@example.com", "9876543210", "pass1234"))
print(register_user("invalidemail", "123", "weak"))
````
