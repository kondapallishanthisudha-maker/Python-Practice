## map() Function in Python

`map()` applies a given function to **every item** in an iterable (list, tuple, etc.) and returns a map object (which you usually convert to a list).

**Syntax:**
```
map(function, iterable)
```

- `function` can be a normal function or a `lambda`
- `iterable` can be a list, tuple, string, etc.
- `map()` is lazy - it doesn't compute the results until you convert it (e.g. with `list()`)

## Basic map() usage

````python
# Basic map() with a lambda - squaring numbers
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, numbers))
print(squared)
````

````python
# map() with a normal function instead of lambda
def cube(x):
    return x ** 3

numbers = [1, 2, 3, 4]
cubed = list(map(cube, numbers))
print(cubed)
````

````python
# map() to double every number
numbers = [10, 20, 30, 40]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
````

````python
# map() checking the return type before converting
numbers = [1, 2, 3]
result = map(lambda x: x + 1, numbers)
print(result)          # map object
print(list(result))    # actual values
````

## map() with Strings

````python
# map() to convert a list of strings to uppercase
words = ["apple", "banana", "cherry"]
upper_words = list(map(lambda w: w.upper(), words))
print(upper_words)
````

````python
# map() to capitalize a list of names
names = ["riya", "aman", "kabir"]
capitalized = list(map(lambda name: name.capitalize(), names))
print(capitalized)
````

````python
# map() to get the length of each word in a list
words = ["cat", "elephant", "dog", "giraffe"]
lengths = list(map(len, words))
print(lengths)
````

````python
# map() to strip whitespace from a list of strings
messy = ["  hello  ", " world ", "  python"]
cleaned = list(map(lambda s: s.strip(), messy))
print(cleaned)
````

## map() with Type Conversion

````python
# map() converting a list of strings to integers
str_nums = ["10", "20", "30"]
int_nums = list(map(int, str_nums))
print(int_nums)
````

````python
# map() converting a list of numbers to strings
numbers = [1, 2, 3, 4]
str_numbers = list(map(str, numbers))
print(str_numbers)
print(", ".join(str_numbers))
````

````python
# map() converting user input (split string) into a list of ints
user_input = "5 10 15 20"
numbers = list(map(int, user_input.split()))
print(numbers)
````

````python
# map() converting a list of floats to rounded integers
prices = [19.99, 45.5, 100.75]
rounded = list(map(round, prices))
print(rounded)
````

## map() with Multiple Iterables

````python
# map() with two lists - adding corresponding elements
list1 = [1, 2, 3]
list2 = [10, 20, 30]
summed = list(map(lambda x, y: x + y, list1, list2))
print(summed)
````

````python
# map() with three lists at once
a = [1, 2, 3]
b = [4, 5, 6]
c = [7, 8, 9]
result = list(map(lambda x, y, z: x + y + z, a, b, c))
print(result)
````

````python
#  map() multiplying prices by quantities from two separate lists
prices = [50, 20, 80]
quantities = [2, 5, 1]
totals = list(map(lambda p, q: p * q, prices, quantities))
print(totals)
````

## map() with Dictionaries and Complex Data

````python
#  map() extracting a specific key from a list of dictionaries
students = [{"name": "Riya", "marks": 88}, {"name": "Aman", "marks": 76}]
names_only = list(map(lambda s: s["name"], students))
print(names_only)
````

````python
# map() applying a calculation on dictionary values
students = [{"name": "Riya", "marks": 88}, {"name": "Aman", "marks": 76}]
bonus_marks = list(map(lambda s: s["marks"] + 5, students))
print(bonus_marks)
````

````python
# map() applied to a list of tuples
points = [(1, 2), (3, 4), (5, 6)]
sums = list(map(lambda p: p[0] + p[1], points))
print(sums)
````

## map() vs List Comprehension

````python
# map() and list comprehension doing the same thing
numbers = [1, 2, 3, 4, 5]

map_result = list(map(lambda x: x ** 2, numbers))
comprehension_result = [x ** 2 for x in numbers]

print(map_result)
print(comprehension_result)
print(map_result == comprehension_result)
````

## Combining map() with Other Functions

````python
# map() combined with filter() - square only the even numbers
numbers = list(range(1, 11))
result = list(map(lambda x: x ** 2, filter(lambda x: x % 2 == 0, numbers)))
print(result)
````

````python
# map() combined with sorted() - sort words by their mapped lengths
words = ["banana", "kiwi", "apple", "fig"]
lengths = list(map(len, words))
print(lengths)
print(sorted(words, key=len))
````

````python
# map() with sum() to get a total directly
prices = ["100", "250", "75"]
total = sum(map(int, prices))
print(total)
````

## Bigger Examples

````python
# Temperature converter for a whole list using map()
celsius_temps = [0, 20, 37, 100]
fahrenheit_temps = list(map(lambda c: (c * 9/5) + 32, celsius_temps))

for c, f in zip(celsius_temps, fahrenheit_temps):
    print(f"{c}°C = {f}°F")
````

````python
# Applying a discount to a list of prices using map()
prices = [500, 1200, 750, 2000]
discounted = list(map(lambda p: p - (p * 0.10), prices))
print(discounted)
````

````python
# Bigger: processing a CSV-like row of raw string data with map()
raw_row = "Riya,22,88.5"
name, age, marks = raw_row.split(",")

# Convert age and marks to correct types using map with a list of functions
converters = [str, int, float]
values = [name, age, marks]
converted = list(map(lambda f, v: f(v), converters, values))

print(converted)
print([type(v) for v in converted])
````
