## Conditional Statements in Python

Conditional statements let your program make decisions - running different code depending on whether a condition is True or False.

1. **if** - runs a block only if the condition is True
2. **elif** - checks another condition if the previous one was False
3. **else** - runs when none of the above conditions are True

```python
# Basic if
age = 20
if age >= 18:
    print("You are an adult")
```

```python
# if-else
age = 15
if age >= 18:
    print("You are an adult")
else:
    print("You are a minor")
```

```python
# if-elif-else
marks = 75
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

```python
# Nested if
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Too young to enter")
```

```python
# if with logical operator (and)
age = 25
income = 50000
if age >= 18 and income >= 20000:
    print("Eligible for loan")
```

```python
# if with logical operator (or)
day = "Saturday"
if day == "Saturday" or day == "Sunday":
    print("It's the weekend!")
```

```python
# if with not
is_raining = False
if not is_raining:
    print("You can go outside")
```

```python
# Checking multiple conditions with an elif chain
temperature = 35
if temperature > 40:
    print("Very hot")
elif temperature > 30:
    print("Hot")
elif temperature > 20:
    print("Pleasant")
else:
    print("Cold")
```

```python
# Ternary operator - a one-line if-else
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)
```

```python
# Checking membership inside an if
fruit = "mango"
basket = ["apple", "banana", "mango"]
if fruit in basket:
    print("Fruit is in the basket")
```

```python
# Checking type inside an if
value = "25"
if isinstance(value, str):
    print("This is a string")
```

```python
# Chained comparison in if
age = 25
if 18 <= age <= 60:
    print("Working age")
```

```python
# if checking for empty/falsy values
name = ""
if not name:
    print("Name cannot be empty")
```

```python
# if-elif-else with input
choice = input("Enter A, B, or C: ").upper()
if choice == "A":
    print("You chose A")
elif choice == "B":
    print("You chose B")
elif choice == "C":
    print("You chose C")
else:
    print("Invalid choice")
```

```python
# Nested if-else with multiple conditions - login check
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    if password == "1234":
        print("Login successful")
    else:
        print("Incorrect password")
else:
    print("User not found")
```

## Bigger Examples

```python
# Even/odd and positive/negative checker
num = int(input("Enter a number: "))

if num == 0:
    print("Zero")
elif num > 0:
    if num % 2 == 0:
        print("Positive even number")
    else:
        print("Positive odd number")
else:
    if num % 2 == 0:
        print("Negative even number")
    else:
        print("Negative odd number")
```

```python
# Simple grading system
marks = int(input("Enter your marks: "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 40:
    print("Grade: D")
else:
    print("Grade: F")
```

```python
# ATM PIN check - real life example
correct_pin = "1234"
entered_pin = input("Enter your PIN: ")

if entered_pin == correct_pin:
    print("Access granted")
else:
    print("Incorrect PIN")
```

```python
# Leap year checker
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```

```python
# Simple traffic light simulator
signal = input("Enter signal color (red/yellow/green): ").lower()

if signal == "red":
    print("Stop")
elif signal == "yellow":
    print("Get ready")
elif signal == "green":
    print("Go")
else:
    print("Invalid signal")
```

```python
# Checking multiple independent conditions 
num = 15
if num > 10:
    print("Greater than 10")
if num % 5 == 0:
    print("Divisible by 5")    
```

```python
# if with a function call as the condition
def is_valid(age):
    return age >= 18

age = 21
if is_valid(age):
    print("Valid age")
```

```python
# Conditional with multiple or conditions

fruits = "kiwi"
if fruit == "apple" or fruit == "banana" or fruit == "kiwi":
    print("Available fruit")
```

```python
# checking string content with methods

password = "Abc12345"
if password.isalnum() and len(password) >= 8:
    print("Password format looks valid")
```

```python
# Nested & compound conditions
age = 25
citizen = True
registered = True

if age >= 18:
    if citizen:
        if registered:
            print("Eligible to vote")
        else:
            print("Not registered")
    else:
        print("Not a citizen")
else:
    print("Underage")                
```

```python
# combining and/or with parentheses for clarity
age = 17
has_parent_consent = True

if (age >= 18) or (age >= 13 and has_parent_consent):
    print("Can create account")
else:
    print("Cannot create account")    
```

```python
# Validating a range with elif chain
score = 105
if score < 0 :
    print("Invalid: negative score")
elif score > 100:
    print("Invalid: exceeds maximum")
else:
    print("Valid score")        
```

```python
# ternary variations - Nested ternary(use sparingly - can hurt readability)

marks = 55
result = "A" if marks >= 90 else "B" if marks >= 75 else "C" if marks >= 50 else "F"
print(result)
```

```python
# ternary inside an f-string

stock = 0
print(f"Status: {'In Stock' if stock > 0 else 'Out of Stock'}")
```

```python
# BMI calculator with categories
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in meters: "))
bmi = weight / (height ** 2)

if bmi < 18.5:
    category = "Underweight"
elif bmi < 25:
    category = "Normal"
elif bmi < 30:
    category = "Overweight" 
else:
    category = "Obese"    

print(f"BMI: {bmi:.1f} -> {category}")
```

```python
# Simple discount calculator based on purchase amount

amount = float(input("Enter purchase amount: "))

if amount >= 5000:
    discount = 0.20
elif amount >= 2000:
    discount = 0.10
elif amount >= 500:
    discount = 0.05
else:
    discount = 0

final_amount = amount - (amount * discount)
print(f"Discount: {discount*100}% , final amount: ₹{final_amount:.2f}")                
```

```python
# Ticket pricing based on age category 

age = int(input("Enter age: "))

if age < 5:
    price = 0
elif age <= 12:
    price = 100
elif age <= 60:
    price = 250

else:
    price = 150

print(f"Ticket price: ₹{price}")                
```

```python
# simple password validator with multiples rules 

password = input("Create a password: ")

has_upper = any(char.isupper() for char in password)
has_digit = any(char.isdigit() for char in password)
long_enough = len(password) >= 8

if has_upper and has_digit and long_enough:
    print("Strong password")
elif long_enough:
    print("Medium password - add a number and uppercase letter")
else:
    print("Weak password - must be at least 8 characters")      
```

```python
# Season identifier based on month

month = input("Enter month name: ").lower()

if month in ["decemeber", "january", "februray"]:
    print("Winter")
elif month in ["march", "april", "may"]:
    print("Spring")
elif month in ["june", "july", "august"]:
    print("Summer")
elif month in ["september", "october", "november"]:
    print("Autumn")
else:
    print("Invalid month")  
```

```python
# Movie ticket age + day pricing combo (nested + elif together)

age = int(input("Enter age: "))
day = input("Is it a weekday or weekend? ").lower()

if age < 12:
    base_price = 100
elif age <= 60:
    base_price = 250
else:
    base_price = 150

if day == "weekend":
    final_price = base_price + 50
else:
    final_price = base_price

print(f"Ticket price: ₹{final_price}")
```

## Key things to remember

- Only **one** block runs in an if-elif-else chain - as soon as one condition is True, the rest are skipped
- Indentation matters - Python uses it to define code blocks (no `{}` like other languages)
- Conditions can combine `and`, `or`, `not` for more complex decisions
- Empty strings, `0`, `None`, and empty collections are all "falsy" in an `if` check
