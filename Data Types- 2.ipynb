# DATA TYPES

### In Data Types they describe individual values (a number , text, true or false)
1. int() - Whole Numbers : 1,25,14,-1,-25...
2. float() - Decimal Numbers : 3.14,1.2,-0.5,....
3. str() - Sequence of Characters : "shanthi","s",...
4. bool() - Boolean : true or false
5. complex() - Number with real + Imaginary parts: 3+4j,10+3j,...

### INTEGERS

```python
age = 25
print(age, type(age))
```

```python
temparature = -15
population = 1_40_000_000
print(temparature, type(temparature))
print(population, type(population))
```

```python
a = 10
b = 20

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
```

# COMPLEX

```python
c = 4 + 5j
print(c, type(c))
```

```python
print(c.real, c.imag)
```

```python
c1 = 2+3j
c2 = 4-2j
print(c1 + c2)
print(c1 - c2)
print(c1 * c2)
print(c1 / c2)
```

```python
c3 = 6+3j
print(c3.real, c3.imag)
print(c3.conjugate())
print(c3.conjugate().real, c3.conjugate().imag)
```
