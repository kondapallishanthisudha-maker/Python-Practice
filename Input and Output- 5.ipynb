## Input and Output in Python

Input and Output (I/O) are the basic ways a program interacts with the user.

1. **input()** - Takes data from the user (always returns a string)
2. **print()** - Displays output to the screen

- Input → Data given by the user to the program.
- Output → Information displayed by the program to the user.

## PRINT() - Displaying Output

```python
print("Hello, World!")
```

```python
name = "Shanthi"
age = 22

print("Name:", name, "Age:", age)
```

```python
# sep changes how values are separated
print("apple", "banana", "cherry", sep=", ")
```

```python
# end changes what is printed after the value (default is a new line)
print("Loading", end="...")
print("Done!")
```

```python
# f-strings - the modern way to insert variables into text
name = "Shanthi"
age = 22
print(f"{name} is {age} years old")
```

## INPUT() - Taking Input from the User

- input() always returns a **string**, no matter what the user types.
- If you need a number, you must convert it using int() or float().

```python
name = input("Enter your name: ")
print("Hello,", name)
```

```python
# input() always gives back a string - check the type
age = input("Enter your age: ")
print(age, type(age))
```

```python
# Converting input to int for calculations
age = int(input("Enter your age: "))
print("Next year you will be", age + 1)
```

```python
# Converting input to float
height = float(input("Enter your height in meters: "))
print("Height in cm:", height * 100)
```

```python
# Taking multiple values in one line using split()
data = input("Enter name and age separated by space: ")
name, age = data.split()
age = int(age)
print("Name:", name, "Age:", age)
```

```python
# If the name can have multiple words = Use rsplit():
data = input("Enter name and age separated by space: ")

name, age = data.rsplit(" ", 1)
age = int(age)

print("Name:", name, "Age:", age)
```

## Real-Life Example - ATM Machine

- **Input** → Enter PIN (the PIN you type is the input)
- **Output** → "Transaction Successful" or balance shown on screen (that message is the output)

```python
correct_pin = "1234"
entered_pin = input("Enter PIN: ")

if entered_pin == correct_pin:
    print("Transaction Successful")
else:
    print("Incorrect PIN")
```

```python
# Multiple inputs on separate lines with a summary
subject1 = input("Enter subject 1:")
marks1 = int(input("Enter marks for subject 1: "))
subject2 = input("Enter subject 2: ")
marks2 = int(input("Enter marks for subject 2:"))

print(f"{subject1}: {marks1}, {subject2}: {marks2}")
print(f"Total: {marks1 + marks2}")
```

```python
# Taking a list of numbers from one line
nums = input("Enter numbers separated by space: ").split()
nums = [int(n) for n in nums]
print("Numbers:", nums)
print("Sum:" , sum(nums))
print("Max:", max(nums))
```

```python
# Input with a default fallback using try/except
try:
    quantity = int(input("Enter quantity: "))
except ValueError:
    quantity = 1
    print("Invalid input, defaulting to 1")

print("Quantity:", quantity)        
```

```python
# Input with a default fallback using try/except
try:
    quantity = int(input("Enter quantity: "))
except ValueError:
    quantity = 1
    print("Invalid input, defaulting to 1")

print("Quantity:", quantity)        
```

```python
# Printing without a newline in a loop (building one line)
for i in range(1,6):
    print(i, end=" ")  # end = " " new line
```

```python
# Print returning None (a common confusion point)

result = print("Hello") # print()  → SHOW something
print(result) # None --> print() itself doesnt return a value
#return   → GIVE something back
```

```python
# formatted currency output
price = 1499.5
print(f"Price: ₹{price: }")
```

## Bigger Example - Profile Card Generator

This combines input(), type conversion, and print() with f-strings all together.

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
height = float(input("Enter your height (cm): "))
is_student = input("Are you a student? (yes/no): ").lower() == "yes"

print("\n----- PROFILE -----") #\n next line
print(f"Name    : {name}")
print(f"Age     : {age}")
print(f"Height  : {height} cm")
print(f"Student : {is_student}")
print("--------------------")
```

### F- Strings

```python
# Basic F- string
name = "shanthi"
print(f"Name : {name}")
```

```python
# Embedding an expression directly

a = 5
b = 3
print(f"Sum is {a+b}")
```

```python
# calling a function inside an f-string
name = "shanthi"
print(f"Uppercase: {name.upper()}")
```

```python
name = "SHANTHI"
print(f"Lowecase: {name.lower()}")
```

```python
# Formatting with commas (thousands separator)

population = 14000000
print(f"Population: {population:,}")
```

```python
# Padding  - extra space and alignment  - decides where the value should be placed within space

print(f"{'Name': <10}{'Age': >5}")
print(f"{'Riya': <10}{22: >5}")
```

```python
# Padding numbers with zerors

num = 7
print(f"{num:03}")
```

```python
# percentage format

score = 0.876
print(f"Score: {score:.1%}")
```

```python
name = "Riya"
age = 22
print(f"""
Name: {name}
Age: {age}
""")
```

```python
# f- string with a dictionary
person = {"name": "Aman", "age": 24}
print(f"{person['name']} is {person['age']} years old")
```

```python
# Nested f-strings (formatting inside formatting)
width = 10
value = 42
print(f"{value:>{width}}")   # dynamic width from a variable
```

```python
# receipt generator using f- strings
item = "Notebook"
price = 45.5
quantity = 3
total = price * quantity

print(f"{'Item': <12}{'Qty': >5}{'Price: >10'}{'Total':>10}")
print(f"{item: <12}{quantity: >5}{price: >10.2f}{total: >10.2f}")
```
