# What is Typecasting?

Typecasting (type conversion) means converting one data type into another. Python has two kinds:

1. Implicit — Python converts automatically (no data loss)
2. Explicit — you manually convert using functions like int(), float(), str(), etc.

## Implicity Type Conversion(python does it automatically)

```python
# int automatically becomes float in mixed math
a = 10 #int
b = 2.5 #float
result = a + b
print(result, type(result))
```

```python
# bool automatically becomes int in math
x = True
y = 5
print(x + y)     # 6 -> True behaves as 1
print(type(x + y))
```

```python
# Implicit conversion only works between compatible types
# This will FAIL - Python won't implicitly convert str + int
try:
    result = "5" + 5
except TypeError as e:
    print("Error:", e)
```

## Explicit Type Conversion (Typecasting — you do it manually)

```python
# str - int

age_str = "25"
age = int(age_str)
print(age, type(age))
```

```python
# str - float
price_str = "99.99"
price = float(price_str)
print(price, type(price))
```

```python
# int - str
score = 95
score_str = str(score)
print(score_str, type(score_str))
print("Your score is " + score_str)   # now concatenation works
```

```python
# float - int (truncates, doesnt round!)
pi = 3.99
whole = int(pi)
print(whole)    # 3, not 4 -> int() just drops the decimal
```

```python
# int - float 
num = 10
num_float = float(num)
print(num_float, type(num_float))
```

```python
# str - bool
print(bool("hello"))   # True
print(bool(""))        # False -> only empty string is False
print(bool("False"))   # True! -> the STRING "False" is still non-empty
```

```python
# int/float - bool
print(bool(0))      # False
print(bool(1))      # True
print(bool(-5))     # True -> any non-zero number is True
print(bool(0.0))    # False
```

```python
# converting invalid strings raise an error 
try:
    num = int("hello")
except ValueError as e:
    print("Conversion failed:", e)
```

```python
# Converting a string with decimal to int directly fails
try:
    num = int("25.5")   # this fails!
except ValueError as e:
    print("Error:", e)

# Correct way - convert to float first, then int
num = int(float("25.5"))
print(num)   # 25
```

```python
my_list = [1, 2, 2, 3]
my_tuple = tuple(my_list)
my_set = set(my_list)      # removes duplicates
back_to_list = list(my_set)

print(my_tuple, my_set, back_to_list)
```

```python
# str - list
word = "hello"
char_list = list(word)
print(char_list)    # ['h', 'e', 'l', 'l', 'o']
```

```python
# list of characters to strng
chars = ['h', 'e', 'l', 'l', 'o']
word = "".join(chars)
print(word)    # hello
```

```python
# dict - list
person = {"name": "Riya", "age": 22}
keys_list = list(person)
print(keys_list)    
```

```python
# Safe input conversion with validation
user_input = input("Enter your age: ")

if user_input.isdigit():
    age = int(user_input)
    print(f"You are {age} years old")
else:
    print("Please enter a valid number")
```

```python
# Temperature converter (float typecasting in action)
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C is {fahrenheit}°F")
```

```python
# Sum of digits using typecasting (str → list of chars → int)
number = input("Enter a number: ")
digit_sum = sum(int(digit) for digit in number)
print(f"Sum of digits: {digit_sum}")
```

```python
# Bigger: mini bill calculator with full typecasting flow
item = input("Item name: ")
price = float(input("Price per unit: "))          # str -> float
quantity = int(input("Quantity: "))                # str -> int
total = price * quantity

print(f"\n{quantity} x {item} = ₹{total:.2f}")
print(f"Type of total: {type(total)}")
```
