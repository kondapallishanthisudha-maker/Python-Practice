## Libraries in Python - NumPy, Pandas, Matplotlib, Seaborn

These are the four most commonly used **data science libraries** in Python. They are third-party packages, so they must be installed first:
```
pip install numpy pandas matplotlib seaborn
```

1. **NumPy** - fast numerical computing with arrays
2. **Pandas** - working with tabular data (rows and columns), like Excel in Python
3. **Matplotlib** - creating charts and graphs
4. **Seaborn** - built on top of Matplotlib, makes statistical charts easier and prettier

## Key things to remember

- **NumPy** arrays are faster than regular Python lists and support element-wise math operations directly
- **Pandas** `DataFrame` is the main structure - think of it as a table, like an Excel sheet
- **Matplotlib** is the base plotting library - almost everything else (including seaborn) is built on top of it
- **Seaborn** makes statistical plots easier to create and better looking with less code, and works very well directly with pandas DataFrames
- Always call `plt.show()` after building a plot to display it (not always required in notebooks, but good practice)
- Install all of these once using: `pip install numpy pandas matplotlib seaborn`

## NUMPY - Numerical Computing

````python
# Importing numpy and creating a basic array
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)
print(type(arr))
````

````python
# Creating a 2D array (matrix)
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix)
print("Shape:", matrix.shape)
````

````python
# Array creation shortcuts
print(np.zeros(5))          # array of zeros
print(np.ones(5))           # array of ones
print(np.arange(0, 10, 2))  # like range(), but returns an array
print(np.linspace(0, 1, 5)) # 5 evenly spaced values between 0 and 1
````

````python
# Basic array math - operations apply to every element at once
arr = np.array([1, 2, 3, 4, 5])
print(arr + 10)
print(arr * 2)
print(arr ** 2)
````

````python
# Useful statistical functions
arr = np.array([10, 20, 30, 40, 50])
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Max:", np.max(arr))
print("Min:", np.min(arr))
print("Standard Deviation:", np.std(arr))
````

````python
#  Indexing and slicing (works similar to Python lists)
arr = np.array([10, 20, 30, 40, 50])
print(arr[0])       # first element
print(arr[-1])      # last element
print(arr[1:4])     # slice
````

````python
# Reshaping an array
arr = np.arange(1, 13)
reshaped = arr.reshape(3, 4)
print(reshaped)
````

````python
# Boolean filtering - keep only values matching a condition
arr = np.array([5, 12, 8, 20, 3, 17])
print(arr[arr > 10])
````

## PANDAS - Working with Tabular Data

````python
# Creating a Series (a single column of data with labels)
import pandas as pd

marks = pd.Series([88, 92, 76, 95], index=["Riya", "Aman", "Kabir", "Neha"])
print(marks)
````

````python
# Creating a DataFrame (a table with rows and columns) from a dictionary
data = {
    "Name": ["Riya", "Aman", "Kabir"],
    "Age": [22, 24, 21],
    "Marks": [88, 76, 92]
}
df = pd.DataFrame(data)
print(df)
````

````python
#  Viewing basic info about a DataFrame
print(df.head())        # first 5 rows (all 3 here)
print(df.shape)         # (rows, columns)
print(df.columns)       # column names
print(df.dtypes)        # data type of each column
````

````python
# Selecting a single column and multiple columns
print(df["Name"])
print(df[["Name", "Marks"]])
````

````python
# Filtering rows based on a condition
print(df[df["Marks"] >= 80])
````

````python
# Adding a new column
df["Grade"] = ["B", "C", "A"]
print(df)
````

````python
# Sorting a DataFrame by a column
sorted_df = df.sort_values("Marks", ascending=False)
print(sorted_df)
````

````python
# Grouping and aggregating data
data = {
    "Subject": ["Math", "Math", "Science", "Science"],
    "Marks": [80, 90, 70, 85]
}
df2 = pd.DataFrame(data)
print(df2.groupby("Subject")["Marks"].mean())
````

````python
#  Quick summary statistics with describe()
print(df["Marks"].describe())
````

## MATPLOTLIB - Basic Plotting

````python
# Basic line plot
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 50]

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
````

````python
# Bar chart
subjects = ["Math", "Science", "English"]
scores = [85, 90, 78]

plt.bar(subjects, scores, color="skyblue")
plt.title("Subject-wise Scores")
plt.show()
````

````python
# Scatter plot
hours_studied = [1, 2, 3, 4, 5, 6]
marks_scored = [40, 50, 55, 65, 75, 85]

plt.scatter(hours_studied, marks_scored, color="red")
plt.title("Hours Studied vs Marks Scored")
plt.xlabel("Hours Studied")
plt.ylabel("Marks Scored")
plt.show()
````

````python
# Pie chart
labels = ["Rent", "Food", "Savings", "Others"]
amounts = [40, 25, 20, 15]

plt.pie(amounts, labels=labels, autopct="%1.1f%%")
plt.title("Monthly Budget Breakdown")
plt.show()
````

````python
# Multiple lines on the same plot with a legend
days = [1, 2, 3, 4, 5]
product_a = [100, 120, 90, 140, 130]
product_b = [80, 85, 95, 100, 110]

plt.plot(days, product_a, label="Product A")
plt.plot(days, product_b, label="Product B")
plt.title("Sales Comparison")
plt.xlabel("Day")
plt.ylabel("Units Sold")
plt.legend()
plt.show()
````

## SEABORN - Statistical Visualization

````python
# Basic seaborn bar plot using a pandas DataFrame
import seaborn as sns

data = {
    "Subject": ["Math", "Science", "English", "History"],
    "Scores": [85, 90, 78, 72]
}
df = pd.DataFrame(data)

sns.barplot(x="Subject", y="Scores", data=df)
plt.title("Scores by Subject")
plt.show()
````

````python
#Histogram - showing the distribution of values
scores = [55, 60, 65, 70, 70, 75, 80, 80, 80, 85, 90, 95]

sns.histplot(scores, bins=6, kde=True)
plt.title("Distribution of Scores")
plt.show()
````

````python
# Scatter plot with seaborn, colored by category
data = {
    "Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks": [35, 45, 50, 60, 65, 75, 80, 90],
    "Passed": ["No", "No", "No", "Yes", "Yes", "Yes", "Yes", "Yes"]
}
df = pd.DataFrame(data)

sns.scatterplot(x="Hours", y="Marks", hue="Passed", data=df)
plt.title("Hours vs Marks (colored by Pass/Fail)")
plt.show()
````

````python
#  Box plot - useful for spotting outliers and spread of data
data = {
    "Subject": ["Math", "Math", "Math", "Science", "Science", "Science"],
    "Scores": [60, 85, 90, 55, 95, 88]
}
df = pd.DataFrame(data)

sns.boxplot(x="Subject", y="Scores", data=df)
plt.title("Score Spread by Subject")
plt.show()
````

````python
# Heatmap - great for visualizing correlation between numeric columns
data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6],
    "Marks": [35, 45, 55, 65, 75, 85],
    "Attendance": [60, 65, 70, 80, 85, 95]
}
df = pd.DataFrame(data)

sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
````

##  Combining All Four Libraries

````python
# Bigger: analyzing and visualizing student data using numpy + pandas + matplotlib + seaborn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Create the dataset with numpy and pandas
np.random.seed(1)
students = ["Riya", "Aman", "Kabir", "Neha", "Rohan", "Priya"]
marks = np.random.randint(40, 100, size=len(students))

df = pd.DataFrame({"Student": students, "Marks": marks})

# Step 2: Analyze with pandas
print(df)
print("\nAverage marks:", df["Marks"].mean())
print("Top scorer:", df.loc[df["Marks"].idxmax(), "Student"])

# Step 3: Visualize with seaborn/matplotlib
sns.barplot(x="Student", y="Marks", data=df, palette="viridis")
plt.title("Student Marks Overview")
plt.xticks(rotation=30)
plt.show()
````
