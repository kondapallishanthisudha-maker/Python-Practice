## Types of Arguments in Python (Formal and Actual)

When we work with functions, there are two related terms:

1. **Formal Arguments (Parameters)** - the variable names written inside the function definition. They act as placeholders.
2. **Actual Arguments (Arguments)** - the real values passed to the function when it is called.

```python
def greet(name):     # name -> formal argument (parameter)
    print(f"Hello, {name}")

greet("Riya")         # "Riya" -> actual argument (value passed)
```

Python supports different **types of actual arguments** that can be passed while calling a function:
1. Positional Arguments
2. Keyword Arguments
3. Default Arguments
4. Variable-length Arguments (*args)
5. Variable-length Keyword Arguments (**kwargs)

## Key things to remember

- **Formal argument** = the placeholder name in the function definition
- **Actual argument** = the real value passed when calling the function
- **Positional arguments** are matched by order
- **Keyword arguments** are matched by name, so order doesn't matter
- **Default arguments** provide a fallback value if none is passed - they must come after non-default arguments
- **`*args`** collects extra positional actual arguments into a tuple
- **`**kwargs`** collects extra keyword actual arguments into a dictionary
- Correct definition order: `def func(positional, default=value, *args, **kwargs):`

## Formal vs Actual Arguments - Basics

````python
#Formal argument (parameter) vs Actual argument (value passed)
def greet(name):          # 'name' is the FORMAL argument (parameter)
    print(f"Hello, {name}")

greet("Riya")               # "Riya" is the ACTUAL argument (value)
````

````python
# Multiple formal arguments, multiple actual arguments
def add(a, b):             # a, b -> formal arguments
    return a + b

result = add(5, 10)          # 5, 10 -> actual arguments
print(result)
````

````python
# Formal arguments are just names - they only get meaning when the function is called
def multiply(x, y):
    print(f"Formal arguments: x, y")
    return x * y

print(multiply(4, 5))         # actual arguments: 4 and 5
print(multiply(10, 2))        # actual arguments: 10 and 2 (same formal names, different values)
````

## 1. Positional Arguments

Actual arguments are matched to formal arguments based on their **position/order**.

````python
# this is called position argument because name we assigned to position name to nit & age to 28
# how it know that name shanthi & age 21

def person(name, age):
    print(name)
    print(age)

person("shanthi",21)
````

````python
# Basic positional arguments
def student_info(name, age):
    print(f"Name: {name}, Age: {age}")

student_info("Riya", 22)    # position matters: Riya -> name, 22 -> age
````

````python
# Changing the order of positional arguments changes the result
def subtract(a, b):
    return a - b

print(subtract(10, 3))    # 7
print(subtract(3, 10))    # -7  -> order matters for positional arguments
````

````python
# in this case we cant assign name - 21 & age to 'shanthi
def person(name, age):
    print(name)
    print(age)
print(20,'shanthi') # in this case we cant assign name - 20 & age to 'shanthi' them we can assign them as keyword    
````

````python
# Passing wrong number of positional arguments raises an error
def add(a, b):
    return a + b

try:
    print(add(5))   # missing one required argument
except TypeError as e:
    print("Error:", e)
````

## 2. Keyword Arguments

Actual arguments are passed using the **formal argument's name**, so order doesn't matter.

````python
# Basic keyword arguments
def student_info(name, age):
    print(f"Name: {name}, Age: {age}")

student_info(age=22, name="Riya")   # order doesn't matter with keyword arguments
````

````python
# Mixing positional and keyword arguments (positional must come first)
def order_details(item, quantity, price):
    print(f"{quantity} x {item} at ₹{price} each")

order_details("Notebook", quantity=3, price=40)
````

````python
# Keyword arguments must come after positional arguments (this fails otherwise)
def student_info(name, age):
    print(f"Name: {name}, Age: {age}")

try:
    exec('student_info(name="Riya", 22)')   # SyntaxError: positional after keyword
except SyntaxError as e:
    print("Error:", e)
````

## 3. Default Arguments

A formal argument can have a **default value**, used automatically if no actual argument is passed for it.

````python
# Basic default argument
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()          # uses default value "Guest"
greet("Aman")    # overrides default
````

````python
# Multiple default arguments
def student_info(name, age=18, city="Unknown"):
    print(f"Name: {name}, Age: {age}, City: {city}")

student_info("Riya")
student_info("Aman", 22)
student_info("Kabir", 25, "Hyderabad")
````

````python
# Default arguments must come AFTER non-default arguments
# This is INVALID: def func(a=5, b):   -> SyntaxError
def func(a, b=5):   # correct order - non-default first, then default
    return a + b

print(func(10))       # b uses default -> 15
print(func(10, 20))   # b overridden -> 30
````

````python
#  Overriding only one default argument using keyword syntax
def order(item="Pen", quantity=1, price=10):
    total = quantity * price
    print(f"{quantity} x {item} = ₹{total}")

order(price=50)   # only override price, rest use defaults
````

## 4. Variable-length Arguments (*args)

Used when you don't know in advance how many **positional** actual arguments will be passed. They are collected into a tuple.

````python
#  Basic *args - accepts any number of positional arguments
def total_sum(*numbers):
    print(numbers, type(numbers))   # numbers becomes a tuple
    return sum(numbers)

print(total_sum(1, 2, 3))
print(total_sum(10, 20, 30, 40))
````

````python
# *args combined with a regular formal argument
def show_scores(student, *scores):
    print(f"{student}'s scores: {scores}")
    print(f"Average: {sum(scores) / len(scores):.2f}")

show_scores("Riya", 88, 92, 79)
````

````python
# Unpacking a list into *args using the * operator while calling
def add_all(*numbers):
    return sum(numbers)

values = [10, 20, 30]
print(add_all(*values))   # unpacks the list into separate actual arguments
````

## 5. Variable-length Keyword Arguments (**kwargs)

Used when you don't know in advance how many **keyword** actual arguments will be passed. They are collected into a dictionary.

````python
# Basic **kwargs - accepts any number of keyword arguments
def student_profile(**details):
    print(details, type(details))   # details becomes a dictionary
    for key, value in details.items():
        print(f"{key}: {value}")

student_profile(name="Riya", age=22, city="Hyderabad")
````

````python
# **kwargs combined with a regular formal argument
def order_receipt(customer, **items):
    print(f"Customer: {customer}")
    for item, price in items.items():
        print(f"{item}: ₹{price}")

order_receipt("Aman", notebook=40, pen=10, bag=500)
````

````python
# Unpacking a dictionary into **kwargs using ** while calling
def student_info(name, age, city):
    print(f"Name: {name}, Age: {age}, City: {city}")

data = {"name": "Kabir", "age": 25, "city": "Delhi"}
student_info(**data)   # unpacks dictionary into keyword arguments
````

## Combining All Argument Types Together

Correct order in a function definition:
```
def func(positional, default=value, *args, **kwargs):
```

````python
#  Combining positional, default, *args, and **kwargs in one function
def order_summary(customer, discount=0, *items, **info):
    print(f"Customer: {customer}")
    print(f"Discount: {discount}%")
    print(f"Items: {items}")
    print(f"Extra info: {info}")

order_summary("Riya", 10, "Notebook", "Pen", "Bag", payment="UPI", express=True)
````

````python
# Formal vs actual arguments count mismatch with *args (never raises an error)
def flexible_sum(*args):
    return sum(args)

print(flexible_sum())            # 0 actual arguments - works fine, returns 0
print(flexible_sum(5))           # 1 actual argument
print(flexible_sum(5, 10, 15))   # 3 actual arguments - all valid
````

## Bigger Examples

````python
# Bigger: flexible calculator using *args
def calculate(operation, *numbers):
    if operation == "add":
        return sum(numbers)
    elif operation == "multiply":
        result = 1
        for n in numbers:
            result *= n
        return result
    else:
        return "Invalid operation"

print(calculate("add", 1, 2, 3, 4))
print(calculate("multiply", 2, 3, 4))
````

````python
# Bigger: building a user profile card using **kwargs
def build_profile(**fields):
    print("----- Profile -----")
    for key, value in fields.items():
        print(f"{key.capitalize()}: {value}")
    print("--------------------")

build_profile(name="Shanthi", age=22, course="Python", city="Hyderabad")
````

````python
# Bigger: function accepting a mix of all argument types realistically
def book_ticket(passenger, seat_type="Economy", *extra_passengers, **preferences):
    print(f"Main passenger: {passenger}")
    print(f"Seat type: {seat_type}")
    if extra_passengers:
        print(f"Extra passengers: {extra_passengers}")
    if preferences:
        print(f"Preferences: {preferences}")

book_ticket("Riya", "Business", "Aman", "Kabir", meal="Veg", window_seat=True)
````
