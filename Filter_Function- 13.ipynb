## filter() Function in Python

`filter()` builds a new sequence containing only the items from an iterable for which a function returns `True`. It filters out anything the function returns `False` for.

**Syntax:**
```
filter(function, iterable)
```

- `function` must return `True` or `False` for each item (often a `lambda`)
- `iterable` can be a list, tuple, string, etc.
- `filter()` is lazy - it doesn't compute results until you convert it (e.g. with `list()`)
- If `function` is `None`, `filter()` simply removes all falsy values (0, '', None, False, etc.)

## Key things to remember

- `filter()` keeps items where the function returns `True`, and drops the rest
- The filtering function must return a boolean (`True`/`False`) for each item
- `filter(None, iterable)` is a shortcut to remove all falsy values (0, '', None, False, empty collections)
- `filter()` is lazy - always wrap it in `list()`, `sorted()`, `sum()`, etc. to see or use the actual results
- A list comprehension with an `if` condition often does the same job and can be more readable

## Basic filter() usage

````python
# Basic filter() with a lambda - keeping only even numbers
numbers = [10, 15, 22, 33, 40, 55]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)
````

````python
# filter() with a normal function instead of lambda
def is_positive(x):
    return x > 0

numbers = [-5, 3, -2, 8, -1, 10]
positives = list(filter(is_positive, numbers))
print(positives)
````

````python
# filter() to find numbers divisible by both 3 and 5
numbers = list(range(1, 51))
result = list(filter(lambda x: x % 3 == 0 and x % 5 == 0, numbers))
print(result)
````

````python
# filter() checking the return type before converting
numbers = [1, 2, 3, 4, 5]
result = filter(lambda x: x > 2, numbers)
print(result)          # filter object
print(list(result))    # actual values
````

## filter() with None (removing falsy values)

````python
# filter(None, iterable) removes all falsy values automatically
values = [0, 1, "", "hello", None, [], [1, 2], False, True]
cleaned = list(filter(None, values))
print(cleaned)
````

````python
# filter(None, ...) to remove empty strings from a list
data = ["apple", "", "banana", "", "cherry"]
cleaned = list(filter(None, data))
print(cleaned)
````

## filter() with Strings

````python
# filter() to get words longer than 4 letters
words = ["cat", "elephant", "dog", "giraffe", "ox"]
long_words = list(filter(lambda w: len(w) > 4, words))
print(long_words)
````

````python
# filter() combined with a lambda checking short string length
words = ["a", "hello", "hi", "python", "ok"]
short_words = list(filter(lambda w: len(w) <= 2, words))
print(short_words)
````

````python
# filter() to keep only words starting with a specific letter
names = ["Riya", "Aman", "Ravi", "Kabir", "Rohan"]
r_names = list(filter(lambda n: n.startswith("R"), names))
print(r_names)
````

````python
# filter() to keep only alphabetic strings (removing ones with numbers/symbols)
data = ["hello", "world123", "python", "abc!", "code"]
alpha_only = list(filter(lambda s: s.isalpha(), data))
print(alpha_only)
````

## filter() with Numbers

````python
# filter() to keep only odd numbers
numbers = list(range(1, 21))
odd_numbers = list(filter(lambda x: x % 2 != 0, numbers))
print(odd_numbers)
````

````python
# filter() to keep numbers within a range
numbers = [5, 15, 25, 35, 45, 55]
in_range = list(filter(lambda x: 10 <= x <= 40, numbers))
print(in_range)
````

````python
# filter() to find prime numbers using a helper function
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

numbers = list(range(1, 30))
primes = list(filter(is_prime, numbers))
print(primes)
````

````python
# filter() to remove duplicates from a list while keeping first occurrence
numbers = [1, 2, 2, 3, 4, 4, 5]
seen = set()
unique = list(filter(lambda x: not (x in seen or seen.add(x)), numbers))
print(unique)
````

## filter() with Dictionaries and Complex Data

````python
# filter() to keep only students who passed (marks >= 40)
students = [{"name": "Riya", "marks": 88}, {"name": "Aman", "marks": 32}, {"name": "Kabir", "marks": 55}]
passed = list(filter(lambda s: s["marks"] >= 40, students))
print(passed)
````

````python
# filter() applied to a list of tuples
points = [(1, 2), (3, -4), (5, 6), (-1, -2)]
positive_points = list(filter(lambda p: p[0] > 0 and p[1] > 0, points))
print(positive_points)
````

````python
# filter() to find keys in a dictionary meeting a condition
prices = {"apple": 50, "banana": 20, "mango": 80, "grape": 15}
expensive_items = list(filter(lambda item: prices[item] > 30, prices))
print(expensive_items)
````

## filter() vs List Comprehension

````python
# filter() and list comprehension doing the same thing
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

filter_result = list(filter(lambda x: x % 2 == 0, numbers))
comprehension_result = [x for x in numbers if x % 2 == 0]

print(filter_result)
print(comprehension_result)
print(filter_result == comprehension_result)
````

## Combining filter() with Other Functions

````python
#filter() combined with map() - square only the even numbers
numbers = list(range(1, 11))
result = list(map(lambda x: x ** 2, filter(lambda x: x % 2 == 0, numbers)))
print(result)
````

````python
# filter() combined with sum() to total only matching values
expenses = [200, -50, 300, -20, 150]
total_income = sum(filter(lambda x: x > 0, expenses))
print(total_income)
````

````python
#filter() combined with sorted() - filter then sort the results
words = ["banana", "kiwi", "apple", "fig", "grapefruit"]
long_sorted = sorted(filter(lambda w: len(w) > 3, words))
print(long_sorted)
````

## Bigger Examples

````python
# Bigger: filtering a list of user records for valid ages
users = [
    {"name": "Riya", "age": 22},
    {"name": "Aman", "age": -5},
    {"name": "Kabir", "age": 150},
    {"name": "Neha", "age": 30}
]

valid_users = list(filter(lambda u: 0 < u["age"] < 120, users))
print(valid_users)
````

````python
#Bigger: cleaning a user input form (removing empty/None fields)
form_data = {"name": "Riya", "email": "", "phone": "9876543210", "address": None}

filled_fields = dict(filter(lambda item: item[1], form_data.items()))
print(filled_fields)
````

````python
#Bigger: filtering products in stock and within budget
products = [
    {"name": "Headphones", "price": 1500, "in_stock": True},
    {"name": "Keyboard", "price": 800, "in_stock": False},
    {"name": "Mouse", "price": 400, "in_stock": True},
    {"name": "Monitor", "price": 9000, "in_stock": True}
]

budget = 2000
available_within_budget = list(
    filter(lambda p: p["in_stock"] and p["price"] <= budget, products)
)

for p in available_within_budget:
    print(p["name"], "-", p["price"])
````
