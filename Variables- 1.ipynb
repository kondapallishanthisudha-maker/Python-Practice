* Dynamic vs Static Type

```python
def calculate_total(price, tax):
    return price + (price * tax)

print(calculate_total(100, 0.05))    
```

```python
def  calculate_total(price: float, tax: float):
    return price + (price*tax)

# arg 1 has incomplete type'str' ; expected 'float' so it gives error   
print(calculate_total("hundred",0.05))
```
