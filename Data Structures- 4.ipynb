# DATA STRUCTURES IN PYTHON

#### Definition:
#### A data structure is a way of organizing and storing data
#### so that we can access, use, and modify the data efficiently.


#### Types of Data Structures in Python:
#### 1. List
#### 2. Tuple
#### 3. Set
#### 4. Dictionary



Quick Comparison Table — Python Data Structures
=================================================

Structure   | Ordered | Mutable | Duplicates | Syntax
------------|---------|---------|------------|----------------
List        | Yes     | Yes     | Yes        | []
Tuple       | Yes     | No      | Yes        | ()
Dict        | Yes*    | Yes     | No (keys)  | {key: value}
Set         | No      | Yes     | No         | {} or set()

* Dicts maintain insertion order since Python 3.7+

1. LIST []

####  A list is an ordered, mutable (changeable) collection that allows duplicate values

```python
# Basic List
fruits = ["apple", "banana", "cherry"]
print(fruits)
print(type(fruits)) 
```

```python
# Accessing elements(indexing)
fruits = ["apple", "banana", "cherry"] # fruits = [0, 1, 2] reverse=[-3,-2,-1]
print(fruits[0]) # first item
print(fruits[-1]) # second item
```

```python
# slicing a list 
nums = [10,20,30,40,50]
print(nums[1:4]) # [20,30,40]
print(nums[::-1]) # reversed 
```

```python
# Modifying a list(mutability)
fruits = ["apple","banana","cherry"]
fruits[0] = "kiwi" # replacing in place of apply which has index [0]
print(fruits)
```

```python
# Adding elements

fruits = ["apple","banana"]
fruits.append("cherry") # add at end
fruits.insert(1, "mango") # add at specific position
print(fruits)
```

```python
# Removing elements
fruits = ["apple","banana","cherry"]
fruits.remove("banana") # remove by value
popped = fruits.pop() # add at specific position
print(fruits)
print(popped)
```

```python
# List with mixed data types
mixed = [1, "hello", 3.14, True]
print(mixed)
```

```python
# Nested list (list of list)

matrix = [[1,2,3],
          [4,5,6],
          [7,8,9]]
print(matrix[1][2]) # row1 , coln2 
```

```python
# sorting and revesing
nums = [5,2,8,1,9]
nums.sort()
print(nums)
nums.reverse()
print(nums)
```

```python
# list comprehension
squares = [x**2 for x in range(1,6)]
even_nums = [x for x in range(1,20) if x % 2 == 0]
print(squares)
print(even_nums)
```

```python
# Shopping cart total

cart = [
    {"item":"bread","price":40},
    {"item":"milk","price":60},
    {"item":"eggs","price":90},
]
total = sum(item["price"] for item in cart)
print(f"Total:{total}")
```

```python
# Extending a list with another list

list1 = [1,2,3]
list2 = [4,5,6]
list1.extend(list2)
print(list1)
```

```python
# Copying a list correctly (avoiding reference issue)
original = [1,2,3]
copy1 = original.copy()
copy1.append(4)
print(original) # unaffected
print(copy1)
```

```python
# finding max, min, sum, and length together
scores = [67,89,45,92,78]
print(f"Max :{max(scores)}")
print(f"Min :{min(scores)}")
print(f"Count :{len(scores)}")
print(f"Sum :{sum(scores)}")
print(f"Average :{sum(scores)/len(scores)}")
```

```python
nums = [1, 2, 3]
nums.clear() # empty list
print(nums)
```

```python
nums = [10, 20, 30, 20, 40]
print(nums.index(20)) # first occurance
print(nums.count(20))
```

```python
# enumerate while looping
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(index, fruits)
```

```python
# List unpacking with * 
numbers = [1, 2, 3, 4]
first, *middle, last = numbers
print(first, middle, last)
```

```python
# copy vs reference
original = [1, 2, 3]
reference = original 
copy_of_it = original.copy()

reference.append(4)
print(original)   # changed too! [1, 2, 3, 4]
print(copy_of_it) # unaffected[1, 2, 3]
```

```python
# Removing duplicates while preserving order

nums = [1, 3, 2, 3, 1, 4]
unique = list(dict.fromkeys(nums))
print(unique)
```

```python
# filtering a list with a condition 
nums = [10, 15, 20, 25, 30, 35]
even_only = [n for n in nums if n % 2 == 0]
print(even_only)
```

```python
todo_list = []

todo_list.append("Buy groceries")
todo_list.append("clean house")
todo_list.append("finish assignment")

print("--- To-Do List ---")
for i, task in enumerate(todo_list, start=1):
    print(f"{i}.{task}")

completed = todo_list.pop(0)
print(f"\nCompleted: {completed}")
print("Remaining:", todo_list)
```

2. TUPLE()

### A tuple is an ordered but immutable (cannot be changed) collection.

```python
# Basic tuple
dimensions = (1920,1080)
print(dimensions)
print(type(dimensions))
```

```python
# Accessing elements 
point = (10,20,30)
print(point[0], point[-1])
```

```python
# Immutability check
point = (10,20,30)
try:
    point[0] = 99
except TypeError as e :
    print("Error:" , e) # tuple can't be changed    
```

```python
# Tuple unpacking 
coordinates = (12.5, 77.6)
lat, lon = coordinates
print(lat, lon)
```

```python
# single element tuple 
not_a_tuple = (5)
actual_tuple = (5,)
print(type(not_a_tuple))
print(type(actual_tuple))
```

```python
# Tuples as dictionary keys 
location = {
    (12.9, 77.6): "Bangalore",
    (28.6, 77.2): "Delhi"
}
print(location[(12.9,77.6)])
```

```python
def get_min_max(numbers):
    return (min(numbers), max(numbers))
low, high = get_min_max([4,1,9,2,7])
print(f"Min: {low}, Max: {high}")
```

```python
# Counting occurence with .count()
marks = (85,90,85,70,85,60)
print(marks.count(85))
```

```python
colors = ("red","green","blue","yellow")
print(colors.index("blue"))
```

```python
# Concatenating tuples
t1 = (1,2,3)
t2 = (4,5,6)
t3 = t1+t2
print(t3)
```

```python
# reapting a tuple 
t = (1,2)
repeated = t * 3
print(repeated)
```

```python
# Nested tuple
nested = ((1,2), (3,4), (5,6))
print(nested[1][0])
```

```python
# converting list to tuple and back 
my_list = [1,2,3]
my_tuple = tuple(my_list)
back_to_list = list(my_tuple)
print(my_tuple, back_to_list)
```

```python
# using tuples to swap variables
a,b = 5,10
a,b = b,a
print(a,b)
```

```python
# storing and unpacking multiple student records
students = [
    ("Riya", 22, 88.5),
    ("Aman", 24, 76.0),
]

for name, age, marks in students:
    print(f"{name} (Age: {age}) scored {marks}")
```

```python
# tuple length and basic checks 
t = (10, 20, 30, 40)
print(len(t))
print(30 in t)
print(99 not in t)
```

```python
# iterating with enumerate() to get index + value
colors = ("red", "green", "blue")
for index, color in enumerate(colors):
    print(index, color)
```

```python
# sorting a tuple(returns a new list, since tuple can't be modified)
marks = (85, 60, 92, 74)
sorted_marks = sorted(marks)
print(sorted_marks, type(sorted_marks))
print(marks)
```

```python
# Min, Max, and sum on a tuple
prices = (250, 400, 150, 600)
print(f"Min: {min(prices)}, Max: {max(prices)}, Sum: {sum(prices)}")
```

```python
# nested tuple unpacking 
point = (1, (2,3))
x, (y, z) = point
print(x, y, z)
```

```python
point = (1, (2,3))
point = x, (y, z) 
print(x, y, z)
```

```python
# using * to unpack extra values 
num = (1,2,3,4,5)
first, *middle, last = num
print(first)
print(middle)
print(last)
```

```python
# Immutable uter tuple, but mutable inner list[]
t = (1, 2, [3, 4]) 
t[2].append(5) # this works! the list inside is still mutable
print(t) # (1, 2, [3, 4])
```

3. Dictionary 

### A dictionary in Python is a collection of key-value pairs used to store and organize data.

```python
student = {"name": "Riya", "age" :22, "course": "Python"}
print(student, type(student))
```

```python
# Accessing values
student = {"name": "riya", "age": 22}
print(student["name"])
print(student.get("city", "Not found")) # safe access with default 
```

```python
# Adding and updating 
student = {"name": "riya", "age": 22}
student["age"] = 23 # update 
student["grade"] = "A" # add new key
print(student)
```

```python
# Removing keys 
student = {"name": "Riya", "age": 22, "grade": "A"}
del student["grade"]
print(student)
```

```python
# Looping Through keys, values, items
person = {"name": "Aman", "age": 30, "city": "delhi"}
for key in person:
    print(key, "->", person[key])

for key, value in person.items():
    print(f"{key}: {value}")    
```

```python
# checking exixtenxe of a key '

students = {"name": "riya", "age": 22}
print("name" in student)
print("city" in student)
```

```python
# nested dictionary
students = {
    "s1": {"name": "Riya", "age": 22},
    "s2": {"name": "aman", "age": 24}
}
print(students["s1"]["name"])
```

```python
# dic comprehension
squares = {x: x**2 for x in range(1,6)}
print(squares)
```

```python
# word frequency counter

text = "the quick brown fox jumps over the lazy dog the fox runs"
words = text.split()

frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print(frequency)    
```

```python
# .pop() to remove and get a value

student = {"name": "shanthi", "age": 22, "grade": "A"}
removed = student.pop("grade")

print(student, removed)
```

```python
# .updte() to merge dictionaries

student = {"name": "shanthi", "age": 21}
extra_info = {"city": "hyderabad", "age" : 22}
student.update(extra_info) # age gets overwritten, city gets added
print(student)
```

```python
# .setdefault() - insert only if key doesnt exist
student = {"name": "shanthi"}
student.setdefault("age", 18) # adds since "age" missing
student.setdefault("name", "X") # doesn't overwrite existing

print(student)
```

```python
# sorting a dictionary ny value 
scores = {"shanthi": 88, "sravani": 76, "kabir": 95}
sorted_scores = dict(sorted(scores.items(), key=lambda x: x[1], reverse= True))
print(sorted_scores)
```

```python
# dic with list values, appending inside
groups = {"math": [], "science": []}
groups["math"].append("shanthi")
groups["math"].append("aman")
groups["science"].append("sravani")
print(groups)
```

```python
marks = {"math": [], "social": [], "science": []}
marks["math"].append(90)
marks["social"].append(25)
marks["science"].append(85)
print(marks)
```

```python
# inventory management system

inventory = {
    "apple": {"price": 50, "stock": 100},
    "banana": {"price": 20, "stock": 150}
}
item = "apple"
quantity_sold = 10
inventory[item]["stock"] -= quantity_sold
revenue = quantity_sold * inventory[item]["price"]

print(f"sold {quantity_sold}{item}(s) for ₹{revenue}")
print(f"Remaining stock: {inventory[item]['stock']}")
```

```python
# .key(), .values(), .items() as list conversions

person = {"name": "shanthi", "age" :22}
print(list(person.keys()))
print(list(person.values()))
print(list(person.items()))
```

```python
# Merging dictionaries with | 

d1 = {"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}
merged = d1 | d2
print(merged)
```

4. Set {}

### A set is an unordered collection of unique elements in Python. It does not allow duplicate values.

```python
# Basic set

nums = {1, 2, 2, 3, 3, 3}
print(nums) # remove duplicates
```

```python
# Adding and remove

nums = {1,2,3}
nums.add(4)
nums.remove(2)
print(nums)
```

```python
# set operations — union, intersection, difference

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)   # union
print(a & b)   # intersection
print(a - b)   # difference
print(a ^ b)   # symmetric difference
```

```python
# checking membership
fruits = {"apple", "banana", "cherry"}
print("banana" in fruits)
```

```python
# removing duplicates from a list using a set
nums = [1, 2, 2, 3, 3, 3, 4]
unique_nums = list(set(nums))
print(unique_nums)
```

```python
# Frozenset - immutable version of a set
fs = frozenset([1, 2, 3])
print(fs, type(fs)) # fs.add(4) -- error cant modify a frozenset
```

```python
# bigger : finding common elements between two lists

list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]

common = set(list1) & set(list2)
print(f"common elements : {common}")
```

```python
#mini student record system
students = [
    {"name": "Riya", "age": 22, "marks": [85, 90, 78]},
    {"name": "Aman", "age": 24, "marks": [65, 70, 60]},
]

for student in students:
    avg = sum(student["marks"]) / len(student["marks"])
    status = "Pass" if avg >= 40 else "Fail"
    print(f"{student['name']} (Age {student['age']}) -> Avg: {avg:.1f} -> {status}")
```

```python
# Set comprehension with a condition
nums = {x for x in range(1, 20) if x % 3 == 0}
print(nums)
```

```python
# checking equality of sets(order doesn't matter)
a = {1, 2, 3}
b = {3, 2, 1}
print( a == b) # True -> in set doesnt care about order
```

```python
a = 1, 2, 3
b = 3, 2, 1
print( a == b)
```

```python
# Set operations combined in one expression
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7}
c = {1, 8, 9}

result = (a & b) | c
print(result)
```

```python
# student club membership overlap
coding_club = {"Riya", "Aman", "Kabir", "Neha"}
music_club = {"Aman", "Priya", "Kabir"}
sports_club = {"Riya", "Priya", "Rohan"}

in_all_three = coding_club & music_club & sports_club
in_any_club = coding_club | music_club | sports_club
only_coding = coding_club - music_club - sports_club

print(f"In all 3 clubs: {in_all_three}")
print(f"In at least 1 club: {in_any_club}")
print(f"Only in coding club: {only_coding}")
```
